```markdown
# 💻 Python Hacking Adventure (Pure Python Version) 🎮
This is a pure Python implementation of your terminal adventure game. It includes:
- **Intro Trailer**: Text-based cyber cinematics and simulated matrices.
- **Action System**: Run and breach server gates while measuring your distance in inches.
- **Cyber Settings**: Speed settings, notification preferences, and system modes.
- **Interactive Menu UI**: Simulating buttons and notifications inside standard terminal output.
```

````python
import time
import random
import sys

class CyberRunnerGame:
    def __init__(self):
        self.distance_inches = 0.0
        self.encryption_level = 100
        self.security_tier = 1
        self.typing_speed = 0.03
        self.notifications_enabled = True
        self.game_running = True

    def cyber_print(self, text, color_code="\033[32m", delay=None):
        """Helper to simulate hacker typing animation with terminal colors."""
        if delay is None:
            delay = self.typing_speed
        for char in text:
            sys.stdout.write(color_code + char)
            sys.stdout.flush()
            time.sleep(delay)
        print("\033[0m")

    def show_notification(self, alert_text):
        if self.notifications_enabled:
            border = "=" * (len(alert_text) + 4)
            self.cyber_print(f"\n{border}", "\033[31m", delay=0.005)
            self.cyber_print(f"| {alert_text} |", "\033[31m", delay=0.01)
            self.cyber_print(f"{border}\n", "\033[31m", delay=0.005)

    def play_trailer(self):
        print("\n--- [INITIALIZING CYBER TRAILER] ---")
        self.cyber_print("LOADING SYSTEM FILES ... [100%]", "\033[36m")
        time.sleep(0.5)
        self.cyber_print("In a world where security is an illusion...", "\033[33m", 0.05)
        time.sleep(0.5)
        self.cyber_print("One python developer must outrun the Central AI.", "\033[33m", 0.05)
        time.sleep(0.5)
        self.cyber_print("Every inch matters. The nodes are closing in!", "\033[31m", 0.05)
        time.sleep(0.5)
        self.cyber_print("\n=== PYTHON-HACKING-ADVENTURE ===", "\033[32m", 0.01)
        self.cyber_print("  Get ready to run, hack, and evade.  ", "\033[32m", 0.01)
        print("-------------------------------------\n")

    def run_breach(self):
        self.cyber_print("\nInitiating breach... Player is now RUNNING!", "\033[36m")
        for step in range(3):
            time.sleep(0.8)
            inches_run = random.uniform(8.5, 24.2)
            self.distance_inches += inches_run
            self.encryption_level -= random.randint(5, 15)

            self.cyber_print(f"🏃 Running Corridor {step + 1}: Gained {inches_run:.2f} inches!", "\033[32m")

        if random.random() > 0.4:
            self.show_notification("🔥 FIREWALL BREACH DETECTED! ACTIVATE MANUAL EVASION!")
            self.encryption_level = max(self.encryption_level - 30, 10)
        else:
            self.cyber_print("✅ Node bypassed successfully. Safe for now.", "\033[32m")

    def show_status(self):
        print("\n================ SYSTEM PANEL ================")
        print(f"   🏃 Total Distance Run : {self.distance_inches:.2f} inches")
        print(f"   🔒 Security Threat    : Tier {self.security_tier}")
        print(f"   🛡️ System Encryption   : {max(self.encryption_level, 0)}%")
        print("==============================================")

    def run_settings(self):
        print("\n--- CONFIGURATION CONTROLS ---")
        print("1) Set Hack Typing Speed (Fast/Slow)")
        print("2) Toggle Danger Notifications")
        print("3) Return to Deck")
        choice = input("Select Option: ")
        if choice == "1":
            self.typing_speed = 0.005 if self.typing_speed == 0.03 else 0.03
            print(f"Speed adjusted. Delay set to: {self.typing_speed}s")
        elif choice == "2":
            self.notifications_enabled = not self.notifications_enabled
            print(f"Notifications enabled state: {self.notifications_enabled}")

    def main_loop(self):
        self.cyber_print("Welcome Agent. System fully loaded.", "\033[32m")
        while self.game_running:
            self.show_status()
            print("\n[ BUTTON CONTROL DECK ]")
            print("┌──────────────────┐ ┌──────────────────┐")
            print("│ [1] Play Trailer │ │ [2] Breach Port  │")
            print("└──────────────────┘ └──────────────────┘")
            print("┌──────────────────┐ ┌──────────────────┐")
            print("│ [3] Evade Threat │ │ [4] Settings     │")
            print("└──────────────────┘ └──────────────────┘")
            print("             [5] Shut Down Game")

            action = input("\nPress a button [1-5]: ").strip()

            if action == "1":
                self.play_trailer()
            elif action == "2":
                self.run_breach()
            elif action == "3":
                if self.encryption_level < 80:
                    self.cyber_print("\n🛡️ Executing network bypass loop...", "\033[36m")
                    time.sleep(1)
                    self.encryption_level = 100
                    self.cyber_print("Connection encrypted and trace wiped clean!", "\033[32m")
                else:
                    self.cyber_print("\nNo threat active right now. Grid is clear!", "\033[32m")
            elif action == "4":
                self.run_settings()
            elif action == "5":
                self.cyber_print("\nClosing core connection. Goodbye, Cyber Runner.", "\033[31m")
                self.game_running = False
            else:
                print("Invalid cyber sequence entered!")

# Initialize and play the game
game = CyberRunnerGame()
game.main_loop()
````
